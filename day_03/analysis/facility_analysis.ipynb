*Import Libraries*

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

Read Cleaned Dataset*


```python
df = pd.read_csv("../dataset/cleaned_facility_data.csv")

print("Dataset loaded successfully!")
print("Number of rows:", len(df))
print("Number of columns:", len(df.columns))

df.head()
```

*Pandas — Dataset Inspection*

```python
print("Dataset Information:")
df.info()
```

```python
print("Statistical Summary:")
df.describe()
```

```python
print("Dataset Shape:")
print(df.shape)
```

```python
print("Column Names:")
print(df.columns)
```

*Missing Values*

```python
print("Missing values in each column:")
print(df.isnull().sum())
```

*Duplicate Check*

```python
print("Duplicate rows:")
print(df.duplicated().sum())
```

```python
print("Duplicate facility IDs:")
print(df["facility_id"].duplicated().sum())
```

*NumPy Operations*

```python
cleanliness_scores = np.array(df["cleanliness_score"])

print("Cleanliness scores:")
print(cleanliness_scores)

print("Number of dimensions:", cleanliness_scores.ndim)
print("Shape:", cleanliness_scores.shape)
```

*Indexing & Slicing*

```python
print("First cleanliness score:")
print(cleanliness_scores[0])

print("\nFirst five cleanliness scores:")
print(cleanliness_scores[:5])
```

*Mathematical Operations*

```python
print("Average:", np.mean(cleanliness_scores))
print("Maximum:", np.max(cleanliness_scores))
print("Minimum:", np.min(cleanliness_scores))
print("Total:", np.sum(cleanliness_scores))
```

*Pandas Filtering*

```python
poor_cleanliness = df[df["cleanliness_score"] < 5]

print(poor_cleanliness)
```

```python
high_footfall = df[df["footfall"] > 800]

print(high_footfall)
```

*Sorting*

```python
high_complaints = df.sort_values(
    by="complaints",
    ascending=False
)

print(high_complaints.head(10))
```

*Grouping & Aggregation*

```python
location_analysis = df.groupby("location").agg(
    average_cleanliness=("cleanliness_score", "mean"),
    average_odor=("odor_score", "mean"),
    total_footfall=("footfall", "sum"),
    total_complaints=("complaints", "sum")
).reset_index()

location_analysis
```

*Data Transformation*


```python
df["facility_status"] = np.where(
    df["cleanliness_score"] >= 7,
    "Good",
    "Needs Improvement"
)

df[["facility_id", "cleanliness_score", "facility_status"]].head(10)
```

*Outlier Detection*

```python
Q1 = df["footfall"].quantile(0.25)
Q3 = df["footfall"].quantile(0.75)

IQR = Q3 - Q1

lower_limit = Q1 - 1.5 * IQR
upper_limit = Q3 + 1.5 * IQR

print("Lower limit:", lower_limit)
print("Upper limit:", upper_limit)
```

```python
outliers = df[
    (df["footfall"] < lower_limit) |
    (df["footfall"] > upper_limit)
]

print("Footfall outliers:")
print(outliers[["facility_id", "location", "footfall"]])
```

*Useful Statistics*

```python
print("Average cleanliness:",
      df["cleanliness_score"].mean())

print("Average odor score:",
      df["odor_score"].mean())

print("Total footfall:",
      df["footfall"].sum())

print("Total complaints:",
      df["complaints"].sum())

print("Highest complaints:",
      df["complaints"].max())
```

#Visualization 1 — Bar Chart

*Average Cleanliness by Location*

```python
cleanliness_by_location = df.groupby("location")["cleanliness_score"].mean()

plt.figure(figsize=(8, 5))

cleanliness_by_location.plot(kind="bar")

plt.title("Average Cleanliness Score by Location")
plt.xlabel("Location")
plt.ylabel("Average Cleanliness Score")

plt.xticks(rotation=45)
plt.tight_layout()

plt.savefig("../visualizations/01_cleanliness_by_location.png")

plt.show()
```

*Total Complaints by Location*

```python
complaints_by_location = df.groupby("location")["complaints"].sum()

plt.figure(figsize=(8, 5))

complaints_by_location.plot(kind="bar")

plt.title("Total Complaints by Location")
plt.xlabel("Location")
plt.ylabel("Total Complaints")

plt.xticks(rotation=45)
plt.tight_layout()

plt.savefig("../visualizations/02_complaints_by_location.png")

plt.show()
```

#Visualization 2 — Histogram

*Footfall Distribution*

```python
plt.figure(figsize=(8, 5))

plt.hist(df["footfall"], bins=10)

plt.title("Distribution of Facility Footfall")
plt.xlabel("Footfall")
plt.ylabel("Number of Facilities")

plt.tight_layout()

plt.savefig("../visualizations/03_footfall_distribution.png")

plt.show()
```

#Visualization 3 — Scatter Plot

*Footfall vs Complaints*

```python
plt.figure(figsize=(8, 5))

plt.scatter(
    df["footfall"],
    df["complaints"]
)

plt.title("Footfall vs Complaints")
plt.xlabel("Footfall")
plt.ylabel("Complaints")

plt.tight_layout()

plt.savefig("../visualizations/04_footfall_vs_complaints.png")

plt.show()
```

#Visualization 4— Additional Chart

*water availability*

```python
water_count = df["water_availability"].value_counts()

plt.figure(figsize=(6, 6))

plt.pie(
    water_count,
    labels=water_count.index,
    autopct="%1.1f%%"
)

plt.title("Water Availability Across Facilities")

plt.tight_layout()

plt.savefig("../visualizations/05_water_availability.png")

plt.show()
```
